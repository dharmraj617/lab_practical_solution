```python
import pyspark
```

```python
df_airports = spark.read.csv("/Volumes/ourairports/airtport-schema/airportvol/airports.csv", header=True, inferSchema=True)
df_runways = spark.read.csv("/Volumes/ourairports/airtport-schema/airportvol/runways.csv", header=True, inferSchema=True)
```

```python
display(df_airports)
```

```python
display(df_runways)
```

```python
from pyspark.sql.functions import col

df_airports = df_airports.withColumn(
    "latitude_deg",
    col("latitude_deg").cast("double")
).withColumn(
    "longitude_deg",
    col("longitude_deg").cast("double")
).withColumn(
    "elevation_ft",
    col("elevation_ft").cast("integer")
)
```

```python
df_airports.printSchema()
```

```python
df_runways.printSchema()
```

```python
df_joined = df_airports.join(
    df_runways,
    df_airports.ident == df_runways.le_ident,
    "inner"
)
```

```python
display(df_joined)
```

```python
from pyspark.sql.functions import when, col
```

```python
df_joined = df_joined.withColumn(
    "runway_category",
    when(col("length_ft") >= 10000, "Long")
    .when((col("length_ft") >= 5000) & (col("length_ft") < 10000), "Medium")
    .otherwise("Low")
)

display(df_joined)
```

```python
df_joined.createOrReplaceTempView("airport_runway_view")
```

```python
df_joined_1 = df_airports.join(
    df_runways,
    df_airports.id == df_runways.airport_ref,
    "inner"
)

# Create temp view again
df_joined_1.createOrReplaceTempView("airport_runway_view")
```

```python
spark.sql("""
SELECT DISTINCT iso_country
FROM airport_runway_view
ORDER BY iso_country
""").show(300, False)
```

```python
spark.sql("""
SELECT 
    iso_country,
    COUNT(*) AS runway_count
FROM airport_runway_view
GROUP BY iso_country
ORDER BY runway_count DESC
LIMIT 10
""").show()
```

```python
spark.sql("""
SELECT 
    continent,
    AVG(length_ft) AS avg_runway_length
FROM airport_runway_view
WHERE length_ft IS NOT NULL
GROUP BY continent
ORDER BY avg_runway_length DESC
""").show()
```

```python
spark.sql("""
SELECT 
    name,
    municipality,
    COUNT(*) AS lighted_runways
FROM airport_runway_view
WHERE iso_country = 'IN'
AND lighted = 1
GROUP BY name, municipality
HAVING COUNT(*) >= 1
ORDER BY lighted_runways DESC
""").show()
```

```python
spark.sql("""
SELECT DISTINCT iso_country
FROM airport_runway_view
ORDER BY iso_country
""").show(300, False)
```

```python
# Top 5 airports with longest maximum runway length

spark.sql("""
SELECT 
    name,
    iso_country,
    MAX(length_ft) AS max_runway_length
FROM airport_runway_view
WHERE length_ft IS NOT NULL
GROUP BY name, iso_country
ORDER BY max_runway_length DESC
LIMIT 5
""").show()
```
